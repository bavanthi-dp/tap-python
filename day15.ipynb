```python
# task-2

import pandas as pd
import os

data = {}

for file in os.listdir("data"):
    if file.endswith(".csv"):
        data[file] = pd.read_csv("data/" + file)
        print(file, data[file].shape)
```

```python
# task-3

import json

with open("column_mapping.json", "r") as file:
    column_mapping = json.load(file)

print(column_mapping)
```

```python
# task-4

bad_records = []

for file, df in data.items():
    if df.isnull().sum().sum() > 0:
        bad_records.append(file)

        print("Bad file:", file)
        print("Missing values:", df.isnull().sum().sum())

print("Files with missing values:", bad_records)
```

```python
# task-5

class FileError(Exception):
    pass

for file in os.listdir("data"):
    if file.endswith(".csv"):
        try:
            df = pd.read_csv("data/" + file)
            print(file, "loaded successfully")

        except Exception as e:
            raise FileError(f"Failed to read {file}: {e}")
```

```python
# task-6

import logging

logger = logging.getLogger("data")
logger.setLevel(logging.INFO)

handler = logging.FileHandler("logs/process.log")
logger.addHandler(handler)

for file in os.listdir("data"):
    if file.endswith(".csv"):
        logger.info(file + " loaded successfully")

handler.close()
logger.removeHandler(handler)

print("Logging completed")
```

```python
# task-7

import os
import pandas as pd

for file in os.listdir("data"):
    if file.endswith(".csv"):
        df = pd.read_csv("data/" + file)
        df.to_csv("output/" + file, index=False)

print("Processing completed")
```
