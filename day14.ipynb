```python
#task-4
from pathlib import Path
import yaml
import json
import pandas as pd
from day14_pipeline import run_pipeline

# Run the existing pipeline
run_pipeline()

# Read configuration
with open("config.yaml", "r") as file:
    config = yaml.safe_load(file)

output = Path(config["paths"]["output_dir"])

# Read generated files
csv_data = pd.read_csv(output / config["output"]["csv"])
json_data = pd.read_json(output / config["output"]["json"])

print("TASK 4 - CLEAN OUTPUT")
print("\nCSV OUTPUT:")
print(csv_data.head())

print("\nJSON OUTPUT:")
print(json_data.head())

print("\nBoth CSV and JSON files are available!")
```

```python
#task-5
import json
import yaml
from pathlib import Path

# Read configuration
with open("config.yaml", "r") as file:
    config = yaml.safe_load(file)

# Restore original manifest if backup exists
manifest = Path(config["paths"]["manifest"])
backup = Path("manifest_backup.json")

if backup.exists() and not manifest.exists():
    backup.rename(manifest)

# Read the generated summary
summary_file = Path(config["paths"]["output_dir"]) / config["output"]["summary"]

print("TASK 5 - RUN SUMMARY")

with open(summary_file, "r") as file:
    summary = json.load(file)

print(json.dumps(summary, indent=4))
```
