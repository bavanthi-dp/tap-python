```python
#task-1
import pandas as pd
#csv reading function
def read_csv(path):
    df = pd.read_csv(path)
    return df

df = read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\orders.csv")
print(df.head())

#column validation function
def validate_columns(df, required_columns):
    return all(column in df.columns for column in required_columns)

required_columns = ["order_id", "customer_id"]
print(validate_columns(df, required_columns))

#column name transformation function
import re
def to_snake_case(column_name):
    column_name = column_name.strip().lower()
    column_name = re.sub(r"[^a-z0-9]+", "_", column_name)
    return column_name.strip("_")

to_snake_case("Customer Name")
```
