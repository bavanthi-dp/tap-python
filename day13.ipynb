```python

mandatory_fields = {
    "customers.csv": ["customer_id", "customer_unique_id", "customer_zip_code_prefix", "customer_city", "customer_state"],
    "geolocation.csv": ["geolocation_id", "geolocation_zip_code_prefix", "geolocation_lat", "geolocation_lng", "geolocation_city", "geolocation_state"],
    "order_items.csv": ["order_id", "order_item_id", "product_id", "seller_id", "price", "freight_value"],
    "order_payments.csv": ["order_id", "payment_sequential", "payment_type", "payment_installments", "payment_value"],
    "order_reviews.csv": ["review_id", "order_id", "review_score", "review_creation_date", "review_answer_timestamp"],
    "orders.csv": ["order_id", "customer_id", "order_status", "order_purchase_timestamp"],
    "orders_semicolon.csv": ["order_id", "customer_id", "order_status", "order_purchase_timestamp"],
    "product_category_name_translation.csv": ["product_category_name", "product_category_name_english"],
    "products.csv": ["product_id"],
    "sellers.csv": ["seller_id", "seller_zip_code_prefix", "seller_city", "seller_state"]
}

all_errors = []

for file_name in mandatory_fields:

    if file_name == "orders_semicolon.csv":
        df = pd.read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\orders_semicolon.csv", sep=";")
    else:
        df = pd.read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\\" + file_name)

    required_columns = mandatory_fields[file_name]

    invalid_rows = df[df[required_columns].isna().any(axis=1)].copy()

    invalid_rows["file_name"] = file_name

    all_errors.append(invalid_rows)

    print(file_name, "Invalid rows:", len(invalid_rows))

collected_errors = pd.concat(all_errors, ignore_index=True)

print("Total invalid rows:", len(collected_errors))
```

```python
#task-2
all_errors = []

for file_name in mandatory_fields:

    if file_name == "orders_semicolon.csv":
        df = pd.read_csv(
            r"C:\Users\Bavanthika\Desktop\tap-python\data\orders_semicolon.csv",
            sep=";"
        )
    else:
        df = pd.read_csv(
            r"C:\Users\Bavanthika\Desktop\tap-python\data\\" + file_name
        )

    required_columns = mandatory_fields[file_name]

    invalid_rows = df[df[required_columns].isna().any(axis=1)].copy()

    invalid_rows["file_name"] = file_name
    invalid_rows["reason"] = "Missing required value"

    all_errors.append(invalid_rows)

collected_errors = pd.concat(all_errors, ignore_index=True)

collected_errors.to_csv("bad_records.csv", index=False)

print("Total rejected rows:", len(collected_errors))
print("Bad records saved successfully!")
```

```python
# task-4

df = pd.read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\order_items.csv")

invalid_rows = df[
    pd.to_numeric(df["price"], errors="coerce").isna() |
    pd.to_datetime(df["shipping_limit_date"], errors="coerce").isna()
].copy()

invalid_rows["reason"] = "Invalid price or date"

invalid_rows.to_csv("bad_records.csv", index=False)

print("Rejected rows:", len(invalid_rows))
```

```python
# task-5

items = pd.read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\order_items.csv")
orders = pd.read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\orders.csv")

bad_price = items[items["price"] <= 0].copy()
bad_price["reason"] = "Invalid price"

allowed = ["delivered", "shipped", "canceled", "unavailable", "approved", "invoiced", "processing", "created", "payment_review", "pending"]

bad_status = orders[~orders["order_status"].isin(allowed)].copy()
bad_status["reason"] = "Invalid order status"

bad_records = pd.concat([bad_price, bad_status], ignore_index=True)

bad_records.to_csv("bad_records.csv", index=False)

print("Rejected rows:", len(bad_records))
print("Bad records saved successfully!")
```

```python
# task-6

df = pd.read_csv(r"C:\Users\Bavanthika\Desktop\tap-python\data\customers.csv")

bad_zip = df[~df["customer_zip_code_prefix"].astype(str).str.match(r"^\d{5}$")].copy()
bad_zip["reason"] = "Invalid ZIP prefix"

bad_id = df[~df["customer_id"].astype(str).str.match(r"^[a-fA-F0-9]{32}$")].copy()
bad_id["reason"] = "Invalid ID"

bad_records = pd.concat([bad_zip, bad_id], ignore_index=True)

bad_records.to_csv("bad_records.csv", index=False)

print("Rejected rows:", len(bad_records))
```
